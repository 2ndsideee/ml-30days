```python
A=np.arange(12).reshape(3,4)
print(A)
print(A[0])        # строка
print(A[2, 1])     # элемент: строка 2, столбец 1
print(A[:, 0])     # все строки, столбец 0
print(A[-1])       # последняя строка
print(A[0, 1:3])   # срез start:stop, stop не включается
print(A[:, ::3])   # шаг 3
print(A[A % 2 == 0])   # булева маска: только чётные элементы
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

def check_sum(a,b):
    total_sum = 0
    for item in a:
        total_sum += item*b 
    return total_sum
print(a@b)
print(check_sum(a,b))
```

```python
import math
v = np.array([3,4])
def l1_norm(v):
    return (v ** 2)
def l2_norm(v):
    return math.sqrt(sum(x ** 2 for x in  v))
print(l1_norm(v))
print(l2_norm(v))
print(np.linalg.norm(v, ord = 1))
print(np.linalg.norm(v,ord = 2))
```

```python
a = np.array([1,2])
b = np.array([2,4])
def cosine_sum(a,b):
    return a@b/(np.linalg.norm(a) * np.linalg.norm(b))
print(cosine_sum(a,b))
```
